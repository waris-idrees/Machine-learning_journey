# Analyze Numerical Features

```python
num_col= df.select_dtypes(include="number")
num_col
```

```python
for col in num_col:
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
    plt.tight_layout()
plt.show()
```
