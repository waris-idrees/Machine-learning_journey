# Encoding the Columns

### Gender

```python
df["gender"].unique()
```

```python
df["gender"]=df["gender"].map({
    "Male" : 1,
    "Female" : 0,
})
```

### City

```python
df["city"].unique() ## use One hot encoding method
```

```python
df["city"]
```

```python
df = pd.get_dummies(
    df,
    columns=["city"],
    drop_first=True,
    dtype=int
)
```

```python
df.head()
```

```python
df["education"].unique()
```

```python
df = pd.get_dummies(
    df,
    columns=["education"],
    drop_first=True,
    dtype=int
)
```

```python
df.head()
```

```python
df["signup_date"].unique()
```

```python
df["signup_date"] = pd.to_datetime(df["signup_date"])
```

```python
# Check that signup_date exists
print(df["signup_date"].head())

# Convert signup_date to datetime
df["signup_date"] = pd.to_datetime(df["signup_date"])

# Create year feature
df["signup_year"] = df["signup_date"].dt.year

# Show year counts
print(df["signup_year"].value_counts().sort_index())
```

```python
future_dates = df[df["signup_date"] > "2026-08-30"]

print(future_dates.shape)
```

```python
df = df[df["signup_date"] <= "2026-08-30"].copy()
```

```python
print(df["signup_date"].max())
print(df.shape)
```

```python
df["signup_year"] = df["signup_date"].dt.year
df["signup_month"] = df["signup_date"].dt.month
df["signup_day"] = df["signup_date"].dt.day
df["signup_dayofweek"] = df["signup_date"].dt.dayofweek
```

```python
df = df.drop(columns=["signup_date"])
```

```python
df[["signup_year", "signup_month", "signup_day", "signup_dayofweek"]].head()
```

```python
df["has_credit_card"].unique()
```

```python
df["has_credit"] = df["has_credit_card"].str.strip().str.lower()

df["has_credit_card"] = df["has_credit_card"].replace({
    "YES": "Yes",
    "Y": "Yes",
    "yes": "Yes",
    "no" : "No",
    "N" : "No",
    "n" : "No",
    "y" : "Yes"
})
df.value_counts("has_credit")
```

```python
df["has_credit_card"] = df["has_credit_card"].map({
    "Yes": 1,
    "No": 0
})
```

```python
df["employment_type"].unique()
```

```python
df = pd.get_dummies(
    df,
    columns=["employment_type"],
    drop_first=True,
    dtype=int
)
```
